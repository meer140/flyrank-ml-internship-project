## 3. Human review + the no-go list

**Human Review Steps:**
1. Verify topical relevance and current SERP search intent.
2. Audit page structure for missing subtopics, outdated statistics, or broken links.
3. Rewrite title & meta description if CTR is below tier average.

**No-Go List:**
- DO NOT rewrite healthy, top-performing pages (pos 1-3 with stable traffic).
- DO NOT pad thin pages with generic AI fluff without expanding user intent coverage.

```python
print('Human review protocol and No-Go list established.')
```
